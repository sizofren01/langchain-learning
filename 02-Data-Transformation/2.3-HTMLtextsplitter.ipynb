```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string= """""
<!DOCTYPE html>
<html>
<head>
    <title>LangChain HTML Example</title>
</head>
<body>
    <h1>Introduction to AI</h1>
    <p>Artificial Intelligence (AI) is the simulation of human intelligence by machines. 
    It allows computers to perform tasks that normally require human reasoning.</p>

    <div>
        <h2>Applications of AI</h2>
        <p>AI is used in various fields like healthcare, finance, and education. 
        For example, AI can analyze medical data to help doctors diagnose diseases.</p>
    </div>

    <div>
        <h2>Advantages of AI</h2>
        <p>AI can process large amounts of data faster and more accurately than humans.</p>
        <p>It reduces human errors and helps automate repetitive tasks.</p>
    </div>

    <div>
        <h3>Future of AI</h3>
        <p>In the future, AI will continue to grow with more advanced algorithms 
        that understand emotions and context better.</p>
    </div>
</body>
</html>
"""


headers_to_split_on=[
    ("h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3")
]
html_splitter=HTMLHeaderTextSplitter(headers_to_split_on)
html_header_splits=html_splitter.split_text(html_string)
html_header_splits

```
