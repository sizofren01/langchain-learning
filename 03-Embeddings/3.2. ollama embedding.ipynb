1. What Are Ollama Embeddings?

Ollama lets you run models like llama2, mxbai-embed-large, etc., on your local machine — no internet or API key needed.
LangChain can connect to it just like OpenAI or Hugging Face embeddings.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings=(
    OllamaEmbeddings(model="gemma:2b")
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
        "Alpha is first letter of Greek alphabet",
        "Beta is second letter of Greek alphabet"
    ]
)
```
