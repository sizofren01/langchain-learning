sentence transformer


Sentence Transformers is a Python framework that makes it easy to use and train state-of-the-art embedding models for sentences, text, and images. Since late 2023, the library has officially been maintained and hosted by Hugging Face, benefiting from its robust infrastructure and widespread community. 



```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
text="this is a test documents"
query_result=embeddings.embed_query(text)
query_result
```

```python
len(query_result)
```

```python
doc_result=embeddings.embed_documents([text,"Rhis is not a test document"])
doc_result[0]
```
