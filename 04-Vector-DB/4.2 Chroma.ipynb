What is Chroma?

Chroma (also called ChromaDB) is an open-source vector database used to store and search embeddings — just like FAISS.


```python
##building a sample vector
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("speech.txt")
data=loader.load()
data
```

```python
#Split
text_splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=0)
splits=text_splitter.split_documents(data)
```

```python
embedding = OllamaEmbeddings(model="gemma:2b")
vectordb=Chroma.from_documents(documents=splits,embedding=embedding)
vectordb
```

```python
##query it
query="How many champions league won.?"
docs=vectordb.similarity_search(query)
docs[0].page_content
```

```python
#saving in disk
vectordb=Chroma.from_documents(documents=splits,embedding=embedding,persist_directory="./chroma.db")
```

```python
## load from disk
db2=Chroma(persist_directory="./chroma_db",embedding_function=embedding)
docs=db2.similarity_search(query)
print(docs[0].page_content)
```

```python
retriever=vectordb.as_retriever()
retriever.invoke(query)[0].page_content
```
