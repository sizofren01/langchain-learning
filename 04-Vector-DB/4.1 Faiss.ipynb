What is FAISS

FAISS stands for Facebook AI Similarity Search. It is a library developed by Facebook AI for efficient similarity search and clustering of dense vectors.

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

loader=TextLoader("speech.txt")
documents=loader.load()
text_splitter=CharacterTextSplitter(chunk_size=1000,chunk_overlap=30)
docs=text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings = OllamaEmbeddings(model="gemma:2b")
```

```python
db=FAISS.from_documents(docs,embeddings)
db
```

```python
###querying

query="what is nickname of ronaldo?"
docs=db.similarity_search(query)
docs[0].page_content
```

```python
##retriever
retriever=db.as_retriever()
docs=retriever.invoke(query)
docs[0].page_content
```

```python
##Similarity serach with socore

docs_and_sscore=db.similarity_search_with_score(query)
docs_and_sscore
```

```python
embedding_vector=embeddings.embed_query(query)
embedding_vector
```

```python
docs_score=db.similarity_search_by_vector(embedding_vector)
docs_score
```

```python
##saving and loading
db.save_local("faiss_index")
```

```python
new_db=FAISS.load_local("faiss_index",embeddings,allow_dangerous_deserialization=True)
docs=new_db.similarity_search(query)
docs
```
